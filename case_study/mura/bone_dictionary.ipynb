# Bone Dictionary — per-image pixel value summary

100 sample images per class, resized to 224x224 (BILINEAR, aspect ratio ignored — DenseNet-MURA-PyTorch style).
For each sampled image, prints the min, max, and median pixel value over the whole resized image.

```python
# -- Imports & Config ---------------------------------------------------------
import glob
import random

import numpy as np
from PIL import Image

DATA_ROOT  = "./mura_downloads"
CLASSES    = ["ELBOW", "FINGER", "FOREARM", "HAND", "HUMERUS", "SHOULDER", "WRIST"]
IMG_SIZE   = 224   # DenseNet-MURA-PyTorch: resize to 224x224, aspect ratio ignored
N_SAMPLES  = 100   # samples per class
BG_PATCH   = 20    # corner patch side (px) used to estimate the background value
BG_THRESH  = 10    # |pixel - bg| > BG_THRESH  =>  counted as bone/foreground
SEED       = 42

random.seed(SEED)
```

```python
# -- Helpers --------------------------------------------------------------------
def load_resized(path, size=IMG_SIZE):
    """Grayscale + resize to size x size (BILINEAR, aspect ratio ignored)."""
    return np.array(Image.open(path).convert("L").resize((size, size), Image.BILINEAR))
```

```python
# -- Sample 100 images/class, resize to 224x224, print pixel value min/max/median --
sample_results = {}  # cls -> list of (path, resized_img, min, max, median)

for cls in CLASSES:
    paths = sorted(glob.glob(f"{DATA_ROOT}/train/XR_{cls}/**/*.png", recursive=True))
    sample_paths = random.sample(paths, min(N_SAMPLES, len(paths)))

    print(f"\n{'=' * 100}")
    print(f"CLASS: {cls}   ({len(sample_paths)}/{len(paths)} images sampled, resized to {IMG_SIZE}x{IMG_SIZE})")
    print(f"{'=' * 100}")
    print(f"{'#':>4}  {'file':40s} {'min':>6} {'max':>6} {'median':>7}")
    print("-" * 100)

    entries = []
    for i, p in enumerate(sample_paths, 1):
        img = load_resized(p)
        fname = "/".join(p.split("/")[-2:])
        mn, mx, med = int(img.min()), int(img.max()), int(np.median(img))

        print(f"{i:>4}  {fname:40s} {mn:>6} {mx:>6} {med:>7}")
        entries.append((p, img, mn, mx, med))

    sample_results[cls] = entries
```

```python
# -- Images whose resized (224x224) min pixel value is 0 --------------------
zero_min = {
    cls: [(p, img) for p, img, mn, mx, med in entries if mn == 0]
    for cls, entries in sample_results.items()
}

for cls in CLASSES:
    imgs = zero_min[cls]
    print(f"\n{cls}: {len(imgs)}/{len(sample_results[cls])} sampled images have min pixel == 0")
    for p, _ in imgs:
        print(f"    {'/'.join(p.split('/')[-2:])}")
```

```python
# -- Visualize the min==0 images (resized 224x224) --------------------------
import matplotlib.pyplot as plt

N_SHOW = min(max((len(v) for v in zero_min.values()), default=0), 100)  # cap columns shown

if N_SHOW == 0:
    print("No sampled images with min pixel == 0.")
else:
    fig, axes = plt.subplots(len(CLASSES), N_SHOW,
                              figsize=(N_SHOW * 1.2, len(CLASSES) * 1.4),
                              squeeze=False)
    fig.subplots_adjust(wspace=0.02, hspace=0.15)

    for row, cls in enumerate(CLASSES):
        imgs = zero_min[cls][:N_SHOW]
        for col in range(N_SHOW):
            ax = axes[row][col]
            ax.axis("off")
            if col < len(imgs):
                _, img = imgs[col]
                ax.imshow(img, cmap="gray", vmin=0, vmax=255)
            if col == 0:
                ax.set_title(f"{cls}  min=0: {len(zero_min[cls])}/{len(sample_results[cls])}",
                              fontsize=7, loc="left", pad=2)

    plt.suptitle(f"Sampled images with min pixel == 0 (resized {IMG_SIZE}x{IMG_SIZE}, up to {N_SHOW}/class shown)",
                 fontsize=11)
    plt.tight_layout()
    plt.savefig("bone_dictionary_zero_min.png", dpi=150, bbox_inches="tight")
    plt.show()
```

```python
# -- Coarse-grained analysis: ToTensor, then chunk into 16-byte blocks ------
# ToTensor: uint8 [0,255] -> float32 [0.0, 1.0]. float32 = 4 bytes/pixel,
# so a 16-byte block = 4 consecutive pixels (matches the XOR-slice convention
# used elsewhere in this project: XOR_ROWS=224, XOR_COLS=56).
# We then look at the BRIGHT chunks -- high mean value -- since in an X-ray,
# dense bone reads brighter than surrounding soft tissue/background.
CHUNK_BYTES       = 16
PIXELS_PER_CHUNK  = CHUNK_BYTES // 4          # float32 = 4 bytes/pixel -> 4 pixels/chunk
CHUNK_ROWS        = IMG_SIZE                  # 224
CHUNK_COLS        = IMG_SIZE // PIXELS_PER_CHUNK  # 224 / 4 = 56
N_CHUNKS          = CHUNK_ROWS * CHUNK_COLS   # 12544
BRIGHT_THRESH     = 0.80                      # chunk mean >= this -> "bright" (~204/255)


def to_tensor(img):
    """uint8 [0,255] -> float32 [0.0, 1.0], matching torchvision ToTensor."""
    return img.astype(np.float32) / 255.0


def chunk_bright_stats(img_f32):
    """Chunk a ToTensor'd (224,224) float32 image into 16-byte (4-pixel) blocks,
    reshaped to (224, 56). Returns per-chunk mean and a bright-chunk mask."""
    chunks = img_f32.reshape(CHUNK_ROWS, CHUNK_COLS, PIXELS_PER_CHUNK)
    chunk_mean  = chunks.mean(axis=2)             # (224, 56)
    bright_mask = chunk_mean >= BRIGHT_THRESH
    return chunk_mean, bright_mask


bright_results = {}  # cls -> list of (path, chunk_mean, bright_mask, bright_pct)

for cls in CLASSES:
    print(f"\n{'=' * 100}")
    print(f"CLASS: {cls}   (ToTensor + 16-byte/4-pixel chunks -> {CHUNK_ROWS}x{CHUNK_COLS} grid, bright >= {BRIGHT_THRESH})")
    print(f"{'=' * 100}")
    print(f"{'#':>4}  {'file':40s} {'bright_chunks':>13} {'bright_%':>9} {'chunk_mean_max':>15}")
    print("-" * 100)

    entries = []
    for i, (p, img, mn, mx, med) in enumerate(sample_results[cls], 1):
        img_f32 = to_tensor(img)
        chunk_mean, bright_mask = chunk_bright_stats(img_f32)
        n_bright   = int(bright_mask.sum())
        bright_pct = n_bright * 100 / N_CHUNKS
        fname = "/".join(p.split("/")[-2:])

        print(f"{i:>4}  {fname:40s} {n_bright:>13} {bright_pct:>8.1f}% {chunk_mean.max():>15.3f}")
        entries.append((p, chunk_mean, bright_mask, bright_pct))

    bright_results[cls] = entries

print(f"\n{'=' * 100}")
print(f"Per-class bright-chunk % summary (mean >= {BRIGHT_THRESH}, across sampled images)")
print(f"{'=' * 100}")
print(f"{'CLASS':10s} {'mean':>7} {'min':>7} {'max':>7}")
for cls in CLASSES:
    pcts = [e[3] for e in bright_results[cls]]
    print(f"{cls:10s} {np.mean(pcts):>6.1f}% {min(pcts):>6.1f}% {max(pcts):>6.1f}%")
```

```python
# -- Sparsity map (16-byte / 4-pixel chunks) for the zero_min images --------
# ToTensor'd chunk -> 1 when all 4 pixels in the chunk equal 0.0 exactly
# (an all-zero, i.e. pure-black, 16-byte block); otherwise -> 0.
def sparsity_map(img):
    chunks = to_tensor(img).reshape(CHUNK_ROWS, CHUNK_COLS, PIXELS_PER_CHUNK)
    return (chunks == 0.0).all(axis=2).astype(np.uint8)   # (224, 56): 1 = all-zero chunk


N_SHOW_SPARSE = min(max((len(v) for v in zero_min.values()), default=0), 100)

if N_SHOW_SPARSE == 0:
    print("No sampled images with min pixel == 0.")
else:
    fig, axes = plt.subplots(len(CLASSES), N_SHOW_SPARSE,
                              figsize=(N_SHOW_SPARSE * 1.2, len(CLASSES) * 2.0),
                              squeeze=False)
    fig.subplots_adjust(wspace=0.02, hspace=0.15)

    for row, cls in enumerate(CLASSES):
        imgs = zero_min[cls][:N_SHOW_SPARSE]
        for col in range(N_SHOW_SPARSE):
            ax = axes[row][col]
            ax.axis("off")
            if col < len(imgs):
                _, img = imgs[col]
                spar = sparsity_map(img)
                ax.imshow(spar, cmap="gray", vmin=0, vmax=1, aspect="auto")
            if col == 0:
                ax.set_title(f"{cls}  min=0: {len(zero_min[cls])}/{len(sample_results[cls])}",
                              fontsize=7, loc="left", pad=2)

    plt.suptitle(f"Sparsity map of min==0 images: 1=all-zero 16-byte(4px) chunk, 0=not "
                 f"({CHUNK_ROWS}x{CHUNK_COLS} grid, up to {N_SHOW_SPARSE}/class shown)", fontsize=11)
    plt.tight_layout()
    plt.savefig("bone_dictionary_zero_min_sparsity.png", dpi=150, bbox_inches="tight")
    plt.show()
```

## XOR sparsity maps: ref-16, ref-32, ref-64 (zero_min images)

Same grid as the cell above, but using the multi-reference XOR convention from
`mura_bodypart_attacker.ipynb`:
- Each **reference page** is a 16-byte pattern (4 float32 pixels all equal to `u8/255.0`).
- A chunk **matches** a reference when all 16 bytes XOR to 0x00.
- The **count map** shown here = number of reference pages that match each chunk  
  (range 0 → N_REF; in practice ≤ 1 for most chunks since refs are distinct u8 values).

```python
# -- XOR sparsity helpers: multi-ref reference pages ------------------------
# ref 값은 mura_bodypart_attacker.ipynb의 density survey 결과를 공유.
# density survey 셀을 먼저 실행하면 xor_density_survey.npy 가 생성됨.
import os
import numpy as np

CHUNK_BYTES_XOR     = 16   # 4 float32 pixels × 4 bytes
DENSITY_SURVEY_PATH = "./xor_density_survey.npy"

def _build_ref_chunks(ref_u8_list):
    """(N_REF, 16) uint8: each row = the 16-byte float32 representation of u8/255."""
    out = []
    for u8 in ref_u8_list:
        f32 = np.array([u8 / 255.0], dtype=np.float32)
        out.append(np.tile(f32.view(np.uint8), 4))   # (16,) bytes
    return np.array(out, dtype=np.uint8)

def xor_count_map(img, ref_chunks):
    """Return (CHUNK_ROWS, CHUNK_COLS) float32: number of ref pages XOR-matching each chunk.
    img: (224, 224) uint8.  ref_chunks: (N_REF, 16) uint8."""
    img_f32 = img.astype(np.float32) / np.float32(255)
    raw     = img_f32.flatten().view(np.uint8)                         # 224*224*4 bytes
    chunks  = raw.reshape(CHUNK_ROWS, CHUNK_COLS, CHUNK_BYTES_XOR)    # (224, 56, 16)
    xor     = chunks[np.newaxis] ^ ref_chunks[:, np.newaxis, np.newaxis, :]  # (N_REF, 224, 56, 16)
    return (xor == 0).all(axis=3).sum(axis=0).astype(np.float32)      # (224, 56)

# --- density survey 로드 → top-N 선택 (mura_bodypart_attacker.ipynb와 동기화) ---
if not os.path.exists(DENSITY_SURVEY_PATH):
    raise FileNotFoundError(
        f"{DENSITY_SURVEY_PATH} not found.\n"
        "mura_bodypart_attacker.ipynb 의 density survey 셀을 먼저 실행하세요."
    )

_hit     = np.load(DENSITY_SURVEY_PATH)                  # (256,) int64 hit counts
_ranked  = np.argsort(-_hit).tolist()                    # u8 sorted by density desc

REF_U8_16      = _ranked[:16]
REF_U8_32      = _ranked[:32]
REF_U8_64_LIST = _ranked[:64]

assert REF_U8_64_LIST[:16] == REF_U8_16, "nested [:16] broken"
assert REF_U8_64_LIST[:32] == REF_U8_32, "nested [:32] broken"

REF_CHUNKS_16 = _build_ref_chunks(REF_U8_16)
REF_CHUNKS_32 = _build_ref_chunks(REF_U8_32)
REF_CHUNKS_64 = _build_ref_chunks(REF_U8_64_LIST)

# sanity: first zero_min ELBOW image
if zero_min["ELBOW"]:
    _cm = xor_count_map(zero_min["ELBOW"][0][1], REF_CHUNKS_64)
    print(f"64-ref count map: shape={_cm.shape}, "
          f"max={_cm.max():.0f}, non-zero chunks={(_cm > 0).sum()}")

print(f"ref-16 : {REF_U8_16}")
print(f"ref-32 : {REF_U8_32}")
print(f"ref-64[:8]: {REF_U8_64_LIST[:8]} ...")
```

```python
# -- Shared plot function for multi-ref XOR sparsity grids ------------------
def plot_xor_count_grid(ref_chunks, ref_label, save_path=None):
    """
    Plots a CLASS × N_SHOW grid of XOR count maps for zero_min images.
    Color = #ref pages matching each chunk (0 = black, N_REF = bright).
    Uses vmax = 1 so any single-ref match is fully visible (most chunks
    match at most 1 ref since refs are distinct u8 values).
    """
    N_SHOW = min(max((len(v) for v in zero_min.values()), default=0), 100)
    if N_SHOW == 0:
        print("No zero_min images to visualize.")
        return

    n_ref = len(ref_chunks)
    fig, axes = plt.subplots(len(CLASSES), N_SHOW,
                              figsize=(N_SHOW * 1.2, len(CLASSES) * 2.0),
                              squeeze=False)
    fig.subplots_adjust(wspace=0.02, hspace=0.15)

    for row, cls in enumerate(CLASSES):
        imgs = zero_min[cls][:N_SHOW]
        for col in range(N_SHOW):
            ax = axes[row][col]
            ax.axis("off")
            if col < len(imgs):
                _, img = imgs[col]
                cmap_img = xor_count_map(img, ref_chunks)
                # vmax=1: any match lights up fully; rare multi-match clips to white
                ax.imshow(cmap_img, cmap="gray", vmin=0, vmax=1, aspect="auto")
            if col == 0:
                ax.set_title(
                    f"{cls}  min=0: {len(zero_min[cls])}/{len(sample_results[cls])}",
                    fontsize=7, loc="left", pad=2)

    plt.suptitle(
        f"XOR sparsity — {ref_label} ({n_ref} reference pages)  "
        f"[{CHUNK_ROWS}×{CHUNK_COLS} chunk grid]\n"
        f"white = ≥1 ref page XOR-matches the chunk  |  "
        f"up to {N_SHOW} zero_min images per class shown",
        fontsize=10)
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
        print(f"Saved: {save_path}")
    plt.show()

print("plot_xor_count_grid defined")
```

```python
# -- 6번째 zero_min: ref 인덱스별 색상으로 sparsity map 표시 + 범례 ---------
# 독립 실행 가능: ref 관련 변수를 셀 내부에서 직접 정의.
# zero_min, CLASSES, DATA_ROOT, plt, np, Image, os, glob 는 앞 셀 필요.
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
from matplotlib.gridspec import GridSpec
import os, pickle

# -- ref 관련 상수 (독립 정의) -----------------------------------------------
_CHUNK_ROWS     = 224
_CHUNK_COLS     = 56
_CHUNK_BYTES    = 16
_DENSITY_PATH   = './xor_density_survey.npy'
CMAP_BASE_N     = 64

def _build_ref_chunks_local(ref_u8_list):
    out = []
    for u8 in ref_u8_list:
        f32 = np.array([u8 / 255.0], dtype=np.float32)
        out.append(np.tile(f32.view(np.uint8), 4))
    return np.array(out, dtype=np.uint8)

_hit    = np.load(_DENSITY_PATH)
_ranked = np.argsort(-_hit).tolist()

_REF_U8_16      = _ranked[:16]
_REF_U8_32      = _ranked[:32]
_REF_U8_64      = _ranked[:64]
_REF_CHUNKS_0   = _build_ref_chunks_local([0])
_REF_CHUNKS_16  = _build_ref_chunks_local(_REF_U8_16)
_REF_CHUNKS_32  = _build_ref_chunks_local(_REF_U8_32)
_REF_CHUNKS_64  = _build_ref_chunks_local(_REF_U8_64)

# -- 함수 -------------------------------------------------------------------
def xor_ref_index_map(img, ref_chunks):
    img_f32 = img.astype(np.float32) / np.float32(255)
    raw     = img_f32.flatten().view(np.uint8)
    chunks  = raw.reshape(_CHUNK_ROWS, _CHUNK_COLS, _CHUNK_BYTES)
    xor     = chunks[np.newaxis] ^ ref_chunks[:, np.newaxis, np.newaxis, :]
    match   = (xor == 0).all(axis=3)
    has_any = match.any(axis=0)
    return np.where(has_any, match.argmax(axis=0), -1).astype(np.int16)

def ref_index_cmap(n_ref):
    colors = [(0,0,0,1)] + [plt.cm.hsv(i/CMAP_BASE_N) for i in range(n_ref)]
    return mcolors.ListedColormap(colors)

def draw_ref_legend(ax, u8_vals, n_leg_cols):
    ax.set_xlim(0,1); ax.set_ylim(0,1); ax.axis('off')
    n = len(u8_vals)
    n_rows = (n + n_leg_cols - 1) // n_leg_cols
    sw = 1.0 / n_leg_cols
    sh = 1.0 / max(n_rows, 1)
    for i, u8v in enumerate(u8_vals):
        ci = i % n_leg_cols
        ri = n_rows - 1 - i // n_leg_cols
        x = ci * sw;  y = ri * sh
        color = plt.cm.hsv(i / CMAP_BASE_N)
        rect = mpatches.FancyBboxPatch(
            (x + sw*0.03, y + sh*0.1), sw*0.38, sh*0.75,
            boxstyle='square,pad=0', facecolor=color, edgecolor='none')
        ax.add_patch(rect)
        ax.text(x + sw*0.45, y + sh*0.5, str(u8v),
                fontsize=5.5, va='center', ha='left', color='black')

# -- 설정 -------------------------------------------------------------------
IMG_IDX = 6
_configs_col = [
    ('original', None,           None),
    ('ref-0',    _REF_CHUNKS_0,  [0]),
    ('ref-16',   _REF_CHUNKS_16, _REF_U8_16),
    ('ref-32',   _REF_CHUNKS_32, _REF_U8_32),
    ('ref-64',   _REF_CHUNKS_64, _REF_U8_64),
]
N_COLS = len(_configs_col)
LEG_H  = 3.0

gs  = GridSpec(len(CLASSES) + 1, N_COLS,
               height_ratios=[2.4]*len(CLASSES) + [LEG_H],
               hspace=0.25, wspace=0.04)
fig = plt.figure(figsize=(N_COLS * 2.0, len(CLASSES) * 2.4 + LEG_H))

# -- 이미지 그리드 ----------------------------------------------------------
for row, cls in enumerate(CLASSES):
    imgs = zero_min[cls]
    if len(imgs) <= IMG_IDX:
        for col in range(N_COLS):
            fig.add_subplot(gs[row, col]).axis('off')
        continue
    _, img = imgs[IMG_IDX]
    for col, (label, ref_chunks, _) in enumerate(_configs_col):
        ax = fig.add_subplot(gs[row, col])
        ax.axis('off')
        if ref_chunks is None:
            ax.imshow(img, cmap='gray', vmin=0, vmax=255, aspect='auto')
        else:
            n_ref   = len(ref_chunks)
            idx_map = xor_ref_index_map(img, ref_chunks)
            display = (idx_map + 1).astype(np.float32)
            ax.imshow(display, cmap=ref_index_cmap(n_ref),
                      vmin=0, vmax=n_ref, aspect='auto')
        if row == 0:
            ax.set_title(label, fontsize=19, pad=3)
        if col == 0:
            ax.set_ylabel(cls, fontsize=8, rotation=0, labelpad=42, va='center')

# -- 범례 행 ---------------------------------------------------------------
LEG_COLS = {1: 1, 16: 4, 32: 8, 64: 8}
for col, (label, ref_chunks, u8_vals) in enumerate(_configs_col):
    ax_leg = fig.add_subplot(gs[len(CLASSES), col])
    if ref_chunks is None or u8_vals is None:
        ax_leg.axis('off')
        continue
    draw_ref_legend(ax_leg, u8_vals, LEG_COLS.get(len(ref_chunks), 8))

plt.savefig('bone_dictionary_compare_ref_colored.pdf', bbox_inches='tight')
plt.savefig('bone_dictionary_compare_ref_colored.png', dpi=150, bbox_inches='tight')
print('Saved PDF + PNG')
plt.show()
```

```python
# -- 전체 데이터셋 16-byte chunk 빈도 분석 (non-identical 포함) ---------------
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
import pickle, time, csv

CHUNK_FREQ_PKL   = './chunk_freq_all.pkl'
CHUNK_FREQ_TSV   = './chunk_freq_all.tsv'
CHUNK_TOP100_CSV = './chunk_freq_top100.csv'

_ref64_set   = set(map(bytes, REF_CHUNKS_64))
_ranked_list = list(_ranked)

_tr_paths = []
for _cls in CLASSES:
    _tr_paths += glob.glob(f'{DATA_ROOT}/train/XR_{_cls}/**/*.png', recursive=True)
print(f'tr_paths: {len(_tr_paths):,} images')

def _count_chunks_one(path):
    arr = np.array(Image.open(path).convert('L').resize((224, 224), Image.BILINEAR),
                   dtype=np.float32) / np.float32(255)
    raw = arr.flatten().view(np.uint8).reshape(-1, 16)
    return Counter(map(bytes, raw))

if not os.path.exists(CHUNK_FREQ_PKL):
    print(f'Scanning {len(_tr_paths):,} train images...')
    t0 = time.time()
    total = Counter()
    WORKERS = min(16, os.cpu_count() or 8)
    done = 0
    with ThreadPoolExecutor(max_workers=WORKERS) as ex:
        futs = [ex.submit(_count_chunks_one, p) for p in _tr_paths]
        for fut in as_completed(futs):
            total += fut.result()
            done += 1
            if done % 5000 == 0:
                print(f'  {done:>6}/{len(_tr_paths)}  unique: {len(total):,}  {time.time()-t0:.0f}s')
    sorted_chunks = total.most_common()
    with open(CHUNK_FREQ_PKL, 'wb') as f:
        pickle.dump(sorted_chunks, f)
    print(f'Done  unique={len(sorted_chunks):,}  ({time.time()-t0:.0f}s)')
else:
    with open(CHUNK_FREQ_PKL, 'rb') as f:
        sorted_chunks = pickle.load(f)
    print(f'Loaded: {CHUNK_FREQ_PKL}  ({len(sorted_chunks):,} unique chunks)')

# -- 통계 -------------------------------------------------------------------
total_n  = sum(c for _, c in sorted_chunks)
n_unique = len(sorted_chunks)
n_single = sum(1 for _, c in sorted_chunks if c == 1)
counts   = np.array([c for _, c in sorted_chunks], dtype=np.int64)

print(f'\n총 청크 수       : {total_n:,}')
print(f'유니크 청크 수   : {n_unique:,}')
print(f'singleton (x1)  : {n_single:,}  ({100*n_single/n_unique:.1f}%)')
print(f'top-1 count     : {counts[0]:,}  ({100*counts[0]/total_n:.4f}%)')
print(f'top-10 cumul    : {counts[:10].sum():,}  ({100*counts[:10].sum()/total_n:.2f}%)')
print(f'top-100 cumul   : {counts[:100].sum():,}  ({100*counts[:100].sum()/total_n:.2f}%)')

def decode_u8(b):
    return tuple(int(x) for x in (np.frombuffer(b, dtype=np.float32) * 255).round())

def ref64_rank(b):
    if b not in _ref64_set:
        return ''
    u8v = decode_u8(b)[0]
    dr = (_ranked_list[:64].index(u8v) + 1) if u8v in _ranked_list[:64] else '?'
    return dr

# -- Top-100 출력 -----------------------------------------------------------
SHOW_N = 100
print(f'\n-- Top-{SHOW_N} -----------------------------------------------------------')
print(f"{'rank':>6}  {'count':>10}  {'cumul%':>7}  {'u8x4':>16}  {'ref64':>6}  hex")
cumul = 0
for rank, (chunk, cnt) in enumerate(sorted_chunks[:SHOW_N], 1):
    cumul += cnt
    u8s = decode_u8(chunk)
    r64 = ref64_rank(chunk)
    marker = f'  <- ref-64 #{r64}' if r64 != '' else ''
    print(f'{rank:>6}  {cnt:>10,}  {100*cumul/total_n:>6.3f}%  '
          f'{str(u8s):>16}  {str(r64):>6}  {chunk.hex()}{marker}')

# -- Bottom-50 출력 ---------------------------------------------------------
print(f'\n-- Bottom-50 --------------------------------------------------------------')
print(f"{'rank':>8}  {'count':>8}  {'u8x4':>16}  hex")
for rank, (chunk, cnt) in enumerate(sorted_chunks[-50:], n_unique - 49):
    u8s = decode_u8(chunk)
    print(f'{rank:>8}  {cnt:>8,}  {str(u8s):>16}  {chunk.hex()}')

# -- Top-100 CSV 저장 -------------------------------------------------------
with open(CHUNK_TOP100_CSV, 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['rank','count','cumul_pct','u8_p0','u8_p1','u8_p2','u8_p3','ref64_rank','hex'])
    cumul = 0
    for rank, (chunk, cnt) in enumerate(sorted_chunks[:SHOW_N], 1):
        cumul += cnt
        u8s = decode_u8(chunk)
        r64 = ref64_rank(chunk)
        w.writerow([rank, cnt, f'{100*cumul/total_n:.4f}',
                    u8s[0], u8s[1], u8s[2], u8s[3], r64, chunk.hex()])
print(f'\nTop-100 CSV saved -> {CHUNK_TOP100_CSV}')

# -- 전체 TSV 저장 ----------------------------------------------------------
print(f'Saving full ranking -> {CHUNK_FREQ_TSV} ...')
with open(CHUNK_FREQ_TSV, 'w') as f:
    f.write('rank\tcount\tu8_p0\tu8_p1\tu8_p2\tu8_p3\thex\tref64_rank\n')
    for rank, (chunk, cnt) in enumerate(sorted_chunks, 1):
        u8s = decode_u8(chunk)
        r64 = ref64_rank(chunk)
        f.write(f'{rank}\t{cnt}\t{u8s[0]}\t{u8s[1]}\t{u8s[2]}\t{u8s[3]}\t{chunk.hex()}\t{r64}\n')
print(f'Saved ({n_unique:,} rows)')
```

```python
# -- ref-64 커버리지 비교: density-based identical vs 전체빈도 top-64 --------
# sorted_chunks (b4c21e97 셀) 로드 결과를 재사용. pkl이 있으면 독립 실행 가능.
import pickle

# -- 데이터 로드 (없으면 pkl에서) -------------------------------------------
if 'sorted_chunks' not in dir() or sorted_chunks is None:
    with open('./chunk_freq_all.pkl', 'rb') as _f:
        sorted_chunks = pickle.load(_f)
    print(f'Loaded chunk_freq_all.pkl  ({len(sorted_chunks):,} unique chunks)')

def _decode_u8(b):
    return tuple(int(x) for x in (np.frombuffer(b, dtype=np.float32) * 255).round())

# -- 두 ref-64 집합 정의 ----------------------------------------------------
# (A) 기존: density survey 기반 identical 청크 top-64
_hit     = np.load('./xor_density_survey.npy')
_ranked  = np.argsort(-_hit).tolist()
_u8_A    = _ranked[:64]
_ref_A   = set()
for _u8 in _u8_A:
    _f32 = np.array([_u8 / 255.0], dtype=np.float32)
    _ref_A.add(bytes(np.tile(_f32.view(np.uint8), 4)))

# (B) 새로운: 전체 빈도 기준 top-64 (identical + non-identical 혼합)
_ref_B = {chunk for chunk, _ in sorted_chunks[:64]}

# -- 전체 커버리지 (재스캔 불필요: sorted_chunks 집계에서 직접 계산) ----------
_total_n = sum(cnt for _, cnt in sorted_chunks)
_cov_A   = sum(cnt for chunk, cnt in sorted_chunks if chunk in _ref_A)
_cov_B   = sum(cnt for chunk, cnt in sorted_chunks if chunk in _ref_B)

print(f'총 청크 수: {_total_n:,}')
print()
print(f'(A) density-based identical top-64  커버리지: '
      f'{_cov_A:>12,}  ({100*_cov_A/_total_n:.3f}%)')
print(f'(B) 전체빈도 top-64 (혼합)          커버리지: '
      f'{_cov_B:>12,}  ({100*_cov_B/_total_n:.3f}%)')
print(f'차이 (B-A): {_cov_B - _cov_A:+,}  ({100*(_cov_B-_cov_A)/_total_n:+.3f}%p)')

# -- 겹침 분석 ---------------------------------------------------------------
_overlap   = _ref_A & _ref_B
_only_A    = _ref_A - _ref_B
_only_B    = _ref_B - _ref_A
print(f'\n겹치는 청크     : {len(_overlap)}/64')
print(f'(A)에만 있는 청크: {len(_only_A)}  (identical이지만 빈도 낮아 B에서 탈락)')
print(f'(B)에 새로 들어온: {len(_only_B)}  (non-identical 또는 더 빈번한 identical)')

# -- 새로 들어온 청크 상세 ---------------------------------------------------
if _only_B:
    print(f'\n(B) 신규 진입 청크 (A 대비):')
    print(f"{'freq_rank':>9}  {'count':>10}  {'type':>10}  u8x4")
    for rank, (chunk, cnt) in enumerate(sorted_chunks[:64], 1):
        if chunk not in _only_B:
            continue
        u8s = _decode_u8(chunk)
        kind = 'identical' if len(set(u8s)) == 1 else 'non-ident'
        print(f'{rank:>9}  {cnt:>10,}  {kind:>10}  {u8s}')

# -- (A)에서 탈락한 청크 상세 ------------------------------------------------
if _only_A:
    print(f'\n(A) 탈락한 identical 청크 (B에서 순위 밀려남):')
    print(f"{'u8':>4}  {'density_rank':>12}  {'freq_count':>10}  {'freq_rank':>9}")
    _chunk_to_rank = {chunk: r for r, (chunk, _) in enumerate(sorted_chunks, 1)}
    for chunk in sorted(_only_A, key=lambda c: _chunk_to_rank.get(c, 999999)):
        u8s = _decode_u8(chunk)
        u8v = u8s[0]
        d_rank = _u8_A.index(u8v) + 1
        f_rank = _chunk_to_rank.get(chunk, '?')
        f_cnt  = sorted_chunks[f_rank-1][1] if isinstance(f_rank, int) else 0
        print(f'{u8v:>4}  {d_rank:>12}  {f_cnt:>10,}  {f_rank:>9}')

# -- 전체 top-64 구성 요약 ---------------------------------------------------
print(f'\n(B) 전체 top-64 구성:')
print(f"{'rank':>5}  {'count':>10}  {'pct%':>6}  {'type':>10}  {'in_A':>5}  u8x4")
_cumul = 0
for rank, (chunk, cnt) in enumerate(sorted_chunks[:64], 1):
    _cumul += cnt
    u8s  = _decode_u8(chunk)
    kind = 'identical' if len(set(u8s)) == 1 else 'non-ident'
    in_a = 'Y' if chunk in _ref_A else ''
    print(f'{rank:>5}  {cnt:>10,}  {100*cnt/_total_n:>5.3f}%  '
          f'{kind:>10}  {in_a:>5}  {u8s}')
print(f'cumul top-64: {100*_cumul/_total_n:.3f}%')
```

```python
# -- 클래스별 top-100 chunk 빈도 분석 ---------------------------------------
# 캐시: chunk_freq_per_class.pkl  /  결과: chunk_freq_per_class_top100.csv
# ref-64 포함 여부: * 표시
import pickle, time, csv
from collections import Counter
from concurrent.futures import ThreadPoolExecutor

CHUNK_PER_CLASS_PKL = './chunk_freq_per_class.pkl'
CHUNK_PER_CLASS_CSV = './chunk_freq_per_class_top100.csv'

# -- ref-64 집합 (독립 정의) ------------------------------------------------
_hit_pc    = np.load('./xor_density_survey.npy')
_ranked_pc = np.argsort(-_hit_pc).tolist()
_ref64_set_pc = set()
for _u8 in _ranked_pc[:64]:
    _f32 = np.array([_u8 / 255.0], dtype=np.float32)
    _ref64_set_pc.add(bytes(np.tile(_f32.view(np.uint8), 4)))

def _dec(b):
    return tuple(int(x) for x in (np.frombuffer(b, dtype=np.float32) * 255).round())

def _count_one(path):
    arr = np.array(Image.open(path).convert('L').resize((224, 224), Image.BILINEAR),
                   dtype=np.float32) / np.float32(255)
    raw = arr.flatten().view(np.uint8).reshape(-1, 16)
    return Counter(map(bytes, raw))

# -- 스캔 또는 캐시 로드 ----------------------------------------------------
if not os.path.exists(CHUNK_PER_CLASS_PKL):
    class_freq = {}
    WORKERS = min(16, os.cpu_count() or 8)
    for cls in CLASSES:
        cls_paths = glob.glob(f'{DATA_ROOT}/train/XR_{cls}/**/*.png', recursive=True)
        t0 = time.time()
        total = Counter()
        with ThreadPoolExecutor(max_workers=WORKERS) as ex:
            for cnt in ex.map(_count_one, cls_paths):
                total += cnt
        class_freq[cls] = total.most_common()
        print(f'{cls:10s}: {len(cls_paths):>6,} imgs  unique={len(class_freq[cls]):,}  '
              f'{time.time()-t0:.0f}s')
    with open(CHUNK_PER_CLASS_PKL, 'wb') as f:
        pickle.dump(class_freq, f)
    print(f'Saved -> {CHUNK_PER_CLASS_PKL}')
else:
    with open(CHUNK_PER_CLASS_PKL, 'rb') as f:
        class_freq = pickle.load(f)
    print(f'Loaded: {CHUNK_PER_CLASS_PKL}')

# -- 출력 + CSV 저장 --------------------------------------------------------
SHOW_N   = 100
csv_rows = []

for cls in CLASSES:
    sc      = class_freq[cls]
    total_c = sum(c for _, c in sc)
    print(f'\n{"="*72}')
    print(f'{cls}  총 청크: {total_c:,}  유니크: {len(sc):,}')
    print(f'{"="*72}')
    print(f"{'rank':>5}  {'count':>10}  {'pct%':>6}  {'ref64':>5}  u8x4")
    for rank, (chunk, cnt) in enumerate(sc[:SHOW_N], 1):
        u8s   = _dec(chunk)
        r64   = '*' if chunk in _ref64_set_pc else ''
        print(f'{rank:>5}  {cnt:>10,}  {100*cnt/total_c:>5.3f}%  {r64:>5}  {u8s}')
        csv_rows.append([cls, rank, cnt, f'{100*cnt/total_c:.4f}',
                         u8s[0], u8s[1], u8s[2], u8s[3], r64, chunk.hex()])

with open(CHUNK_PER_CLASS_CSV, 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['class','rank','count','pct','u8_p0','u8_p1','u8_p2','u8_p3','ref64','hex'])
    w.writerows(csv_rows)
print(f'\nSaved -> {CHUNK_PER_CLASS_CSV}  ({len(csv_rows)} rows)')
```

```python
# -- Verify R=G=B=L for grayscale MURA PNG loaded as RGB ---------------------
import glob, random
import numpy as np
from PIL import Image

random.seed(0)
all_paths = glob.glob(f"{DATA_ROOT}/train/**/*.png", recursive=True)
sample_paths = random.sample(all_paths, 200)

n_fail_rgb  = 0   # R != G or R != B
n_fail_l    = 0   # R != L (grayscale)
n_fail_res  = 0   # resize 후에도 동일한지

for p in sample_paths:
    img_rgb = np.array(Image.open(p).convert('RGB'))   # (H, W, 3) uint8
    img_l   = np.array(Image.open(p).convert('L'))     # (H, W)    uint8
    R, G, B = img_rgb[:, :, 0], img_rgb[:, :, 1], img_rgb[:, :, 2]

    if not (np.array_equal(R, G) and np.array_equal(R, B)):
        n_fail_rgb += 1
        print(f"FAIL R!=G or R!=B: {p}")
    if not np.array_equal(R, img_l):
        n_fail_l += 1
        print(f"FAIL R!=L: {p}")

    # resize 후에도 동일한지
    r_res = np.array(Image.open(p).convert('RGB').resize((224, 224), Image.BILINEAR))
    l_res = np.array(Image.open(p).convert('L').resize((224, 224), Image.BILINEAR))
    if not np.array_equal(r_res[:, :, 0], l_res):
        n_fail_res += 1
        print(f"FAIL after resize R[:,:,0]!=L: {p}")

print(f"검사 이미지: {len(sample_paths)}")
print(f"R==G==B 실패: {n_fail_rgb}")
print(f"R==L 실패:    {n_fail_l}")
print(f"resize 후 R==L 실패: {n_fail_res}")
```
